````python
import ast
from collections import Counter

# Extract and count individual genres
genre_counts = Counter()

for genre_list in titles_clean['genres'].dropna():
    try:
        genres = ast.literal_eval(genre_list)
        genre_counts.update(genres)
    except (ValueError, SyntaxError):
        continue

# Get top 10 genres
top_genres = pd.DataFrame(
    genre_counts.most_common(10),
    columns=['Genre', 'Count']
)

# Visualize
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_genres,
    x='Count',
    y='Genre',
    hue='Genre',
    palette='viridis',
    legend=False
)

plt.title('Top 10 Most Common Genres on Amazon Prime Video')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.tight_layout()
plt.show()
````

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀): There is no statistically significant difference in the distribution of IMDb ratings between Movies and TV Shows on Amazon Prime Video.

Alternative Hypothesis (H₁): There is a statistically significant difference in the distribution of IMDb ratings between Movies and TV Shows on Amazon Prime Video.

#### 2. Perform an appropriate statistical test.

````python
# Hypothesis 1: Compare IMDb ratings between Movies and TV Shows

from scipy.stats import mannwhitneyu

# Separate IMDb ratings by content type
movie_ratings = titles_df.loc[
    titles_df['type'] == 'MOVIE', 'imdb_score'
].dropna()

show_ratings = titles_df.loc[
    titles_df['type'] == 'SHOW', 'imdb_score'
].dropna()

# Perform Mann-Whitney U test
statistic, p_value = mannwhitneyu(
    movie_ratings,
    show_ratings,
    alternative='two-sided'
)

print("Mann-Whitney U Test Results")
print("--------------------------------")
print("Movie observations:", len(movie_ratings))
print("TV Show observations:", len(show_ratings))
print("Movie median IMDb rating:", movie_ratings.median())
print("TV Show median IMDb rating:", show_ratings.median())
print("U Statistic:", statistic)
print("P-value:", p_value)

# Decision at 5% significance level
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis.")
    print("There is a statistically significant difference in IMDb rating distributions.")
else:
    print("Fail to reject the Null Hypothesis.")
    print("There is insufficient evidence of a statistically significant difference.")
````

##### Which statistical test have you done to obtain P-Value?

The Mann-Whitney U test was used to compare the IMDb rating distributions of Movies and TV Shows.

##### Why did you choose the specific statistical test?

The Mann-Whitney U test was selected because the analysis compares two independent groups: Movies and TV Shows. It is a non-parametric test that does not require the assumption of normally distributed data. It is suitable for examining whether the rating distributions of these two groups differ significantly.

**Final conclusion**

The p-value (7.94456 × 10⁻²²⁷) is significantly lower than the 0.05 significance level. Therefore, the null hypothesis is rejected.

````python
# Identify outliers using the IQR method

numerical_cols = ['release_year', 'runtime', 'seasons']

for col in numerical_cols:
    Q1 = titles_df[col].quantile(0.25)
    Q3 = titles_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = titles_df[
        (titles_df[col] < lower) |
        (titles_df[col] > upper)
    ]

    print(f"\n{col}")
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outlier Count:", len(outliers))
````

````python
import re

def remove_urls_and_digits(text):
    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Remove words containing digits
    text = re.sub(r'\b\w*\d+\w*\b', '', text)

    return text

text_df['title'] = text_df['title'].apply(remove_urls_and_digits)
text_df['description'] = text_df['description'].apply(remove_urls_and_digits)

print("URLs and words containing digits removed!")
print(text_df[['title', 'description']].head())
````

#### 5. Removing Stopwords & Removing White spaces

````python
!pip install nltk -q

import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    filtered_words = [
        word for word in words if word not in stop_words
    ]
    return ' '.join(filtered_words)

text_df['title'] = text_df['title'].apply(remove_stopwords)
text_df['description'] = text_df['description'].apply(remove_stopwords)

print("Stopword removal completed!")
print(text_df[['title', 'description']].head())
````

````python
# Remove extra spaces, tabs and newline characters

def remove_extra_whitespace(text):
    return ' '.join(text.split())

text_df['title'] = text_df['title'].apply(remove_extra_whitespace)
text_df['description'] = text_df['description'].apply(remove_extra_whitespace)

print("Extra whitespace removal completed!")
print(text_df[['title', 'description']].head())
````

#### 7. Tokenization

````python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

# Tokenize titles and descriptions
text_df['title_tokens'] = text_df['title'].apply(word_tokenize)
text_df['description_tokens'] = text_df['description'].apply(word_tokenize)

print("Tokenization completed!")

print(text_df[['title_tokens', 'description_tokens']].head())
````

#### 8. Text Normalization

````python
import nltk
import pandas as pd
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

text_df['title_lemmas'] = text_df['title_tokens'].apply(lemmatize_tokens)
text_df['description_lemmas'] = text_df['description_tokens'].apply(lemmatize_tokens)

print("Text normalization using lemmatization completed!")
print(text_df[['title_lemmas', 'description_lemmas']].head())
````

##### Which text normalization technique have you used and why?

Lemmatization was used as the text normalization technique. It converts words into their meaningful dictionary base forms while preserving their linguistic meaning. It was preferred over stemming because stemming may produce incomplete or non-dictionary words. This technique helps maintain consistency in the textual data and improves its suitability for further NLP processing.

#### 9. Part of speech tagging

````python
import nltk

nltk.download('averaged_perceptron_tagger_eng')

# Apply POS tagging to lemmatized tokens
text_df['title_pos'] = text_df['title_lemmas'].apply(
    nltk.pos_tag
)

text_df['description_pos'] = text_df['description_lemmas'].apply(
    nltk.pos_tag
)

print("POS tagging completed!")

print(text_df[['title_pos', 'description_pos']].head())
````

#### 10. Text Vectorization

````python
from sklearn.feature_extraction.text import TfidfVectorizer

# Convert lemmatized tokens back into sentences
text_df['processed_description'] = text_df['description_lemmas'].apply(
    lambda tokens: ' '.join(tokens)
)

# Apply TF-IDF vectorization
tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,
    stop_words='english'
)

tfidf_matrix = tfidf_vectorizer.fit_transform(
    text_df['processed_description']
)

print("TF-IDF vectorization completed!")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Vocabulary size:", len(tfidf_vectorizer.vocabulary_))
````

##### Which text vectorization technique have you used and why?

TF-IDF (Term Frequency–Inverse Document Frequency) was used for text vectorization. It converts textual descriptions into numerical feature vectors by assigning importance to words based on their frequency within a document and across the document collection. TF-IDF helps emphasize informative words while reducing the influence of commonly occurring terms. It was selected because it is a widely used and effective technique for representing textual data in NLP applications.

**Text Vectorization – Result**

TF-IDF vectorization was successfully performed on the preprocessed movie and TV show descriptions. The resulting matrix contains 9,868 documents and 5,000 features, representing the selected vocabulary.

The technique converts textual descriptions into numerical representations based on word importance, making the data suitable for further NLP analysis.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBRegressor
import numpy as np

# Define the XGBoost model
xgb_tuning = XGBRegressor(
    objective='reg:squarederror',
    random_state=42,
    n_jobs=-1
)

# Define hyperparameter search space
param_grid = {
    'n_estimators': [200, 300, 400, 500],
    'max_depth': [3, 4, 5, 6, 7],
    'learning_rate': [0.01, 0.03, 0.05, 0.1],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0]
}

# Randomized Search with 5-fold cross-validation
random_search = RandomizedSearchCV(
    estimator=xgb_tuning,
    param_distributions=param_grid,
    n_iter=15,
    scoring='r2',
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# Fit the search
random_search.fit(X_train, y_train)

# Best model
best_model = random_search.best_estimator_

print("Best Parameters:")
print(random_search.best_params_)

print("\nBest Cross-Validation R2 Score:")
print(random_search.best_score_)
````

````python
# Predict using the tuned model
y_pred_tuned = best_model.predict(X_test)

# Calculate evaluation metrics
mae_tuned = mean_absolute_error(y_test, y_pred_tuned)
mse_tuned = mean_squared_error(y_test, y_pred_tuned)
rmse_tuned = np.sqrt(mse_tuned)
r2_tuned = r2_score(y_test, y_pred_tuned)

print("Tuned XGBoost Performance")
print("-------------------------")
print("MAE:", mae_tuned)
print("MSE:", mse_tuned)
print("RMSE:", rmse_tuned)
print("R2 Score:", r2_tuned)
````
